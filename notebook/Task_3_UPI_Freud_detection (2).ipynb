```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 1. Create Transaction Amount Bins (Buckets)
bins = [0, 500, 1000, 2500, 5000, 10000, 50000]
labels = [
    "< $500",
    "$500 - $1k",
    "$1k - $2.5k",
    "$2.5k - $5k",
    "$5k - $10k",
    "> $10k",
]

df["Amount_Bin"] = pd.cut(
    df["Transaction_Amount"], bins=bins, labels=labels, right=False
)

# 2. Build Cross-tabulation (Counts & Proportions)
ct_counts = pd.crosstab(
    df["Amount_Bin"],
    df["Fraudulent"].map({0: "Non-Fraud (0)", 1: "Fraud (1)"}),
)
ct_percentages = (
    pd.crosstab(
        df["Amount_Bin"],
        df["Fraudulent"].map({0: "Non-Fraud (0)", 1: "Fraud (1)"}),
        normalize="index",
    )
    * 100
)

# 3. Plot Heatmaps
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Heatmap 1: Absolute Transaction Counts
sns.heatmap(
    ct_counts,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=True,
    linewidths=1,
    ax=axes[0],
)
axes[0].set_title(
    "1. Transaction Count by Amount Range", fontsize=12, fontweight="bold"
)
axes[0].set_xlabel("Fraud Status")
axes[0].set_ylabel("Transaction Amount Bins")

# Heatmap 2: Fraud Rate Percentage per Amount Range
sns.heatmap(
    ct_percentages,
    annot=True,
    fmt=".2f",
    cmap="YlOrRd",
    cbar=True,
    linewidths=1,
    ax=axes[1],
)
axes[1].set_title(
    "2. Fraud Rate Percentage (%) by Amount Range",
    fontsize=12,
    fontweight="bold",
)
axes[1].set_xlabel("Fraud Status")
axes[1].set_ylabel("Transaction Amount Bins")

plt.tight_layout()
plt.show()
```

**Phase 5 Hypothesis Testing **

```python
import pandas as pd
from scipy import stats

# Clean data for statistical tests (drop NaN in key testing columns)
df_clean_amt = df.dropna(subset=['Transaction_Amount', 'Fraudulent'])
df_clean_dev = df.dropna(subset=['Device_Used', 'Fraudulent'])

print("=" * 60)
print("HYPOTHESIS TEST 1: Two-Sample T-Test (Amount vs Fraud)")
print("=" * 60)

# Separate amounts into Fraud and Non-Fraud groups
fraud_amt = df_clean_amt[df_clean_amt['Fraudulent'] == 1]['Transaction_Amount']
non_fraud_amt = df_clean_amt[df_clean_amt['Fraudulent'] == 0][
    'Transaction_Amount'
]

# Perform Welch's T-Test (equal_var=False since variances differ)
t_stat, p_val_t = stats.ttest_ind(fraud_amt, non_fraud_amt, equal_var=False)

print(f"Fraud Mean Amount:     ${fraud_amt.mean():.2f}")
print(f"Non-Fraud Mean Amount: ${non_fraud_amt.mean():.2f}")
print(f"T-Statistic:           {t_stat:.4f}")
print(f"p-value:               {p_val_t:.4e}")

if p_val_t < 0.05:
  print("Result: Reject H₀ (Significant difference in amounts)")
else:
  print(
      "Result: Fail to Reject H₀ (No significant difference in average amounts)"
  )

print("\n" + "=" * 60)
print("HYPOTHESIS TEST 2: Chi-Square Test (Device Used vs Fraud)")
print("=" * 60)

# Create contingency table
contingency_table = pd.crosstab(
    df_clean_dev['Device_Used'], df_clean_dev['Fraudulent']
)
chi2_stat, p_val_chi2, dof, expected = stats.chi2_contingency(contingency_table)

print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"Degrees of Freedom:   {dof}")
print(f"p-value:              {p_val_chi2:.4e}")

if p_val_chi2 < 0.05:
  print("Result: Reject H₀ (Device selection is significantly linked to fraud)")
else:
  print("Result: Fail to Reject H₀ (Device selection has no relation to fraud)")
```

**T-test**

```python
import pandas as pd
from scipy import stats

# Clean missing values
df_clean = df.dropna(subset=['Transaction_Amount', 'Fraudulent'])

# Split into Fraud vs Non-Fraud groups
fraud_amt = df_clean[df_clean['Fraudulent'] == 1]['Transaction_Amount']
non_fraud_amt = df_clean[df_clean['Fraudulent'] == 0]['Transaction_Amount']

# Welch's T-Test
t_stat, p_val = stats.ttest_ind(fraud_amt, non_fraud_amt, equal_var=False)

print("=== TWO-SAMPLE T-TEST ===")
print(f"Fraud Mean Amount:     ${fraud_amt.mean():.2f}")
print(f"Non-Fraud Mean Amount: ${non_fraud_amt.mean():.2f}")
print(f"T-Statistic:           {t_stat:.4f}")
print(f"p-value:               {p_val:.4e}")

if p_val < 0.05:
  print("Result: Reject Null Hypothesis (H₀)")
else:
  print("Result: Fail to Reject Null Hypothesis (H₀)")
```

 **Chi-Square Test** ($\chi^2$) (Comparing Categorical Variables)

```python
import pandas as pd
from scipy import stats

# Clean missing values
df_clean_dev = df.dropna(subset=['Device_Used', 'Fraudulent'])

# Contingency Table
contingency_table = pd.crosstab(
    df_clean_dev['Device_Used'], df_clean_dev['Fraudulent']
)

# Chi-Square Test
chi2_stat, p_val_chi2, dof, expected = stats.chi2_contingency(contingency_table)

print("=== CHI-SQUARE TEST OF INDEPENDENCE ===")
print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"Degrees of Freedom:   {dof}")
print(f"p-value:              {p_val_chi2:.4e}")

if p_val_chi2 < 0.05:
  print(
      "Result: Reject Null Hypothesis (H₀) - Device type is significantly linked"
      " to fraud."
  )
else:
  print(
      "Result: Fail to Reject Null Hypothesis (H₀) - Device type is independent"
      " of fraud."
  )
```

**One-Way ANOVA (Comparing Numeric Values Across 3+ Groups)**

```python
import pandas as pd
from scipy import stats

# Clean missing values
df_clean_pay = df.dropna(subset=['Transaction_Amount', 'Payment_Method'])

# Group amounts by Payment Method
groups = [
    group['Transaction_Amount'].values
    for name, group in df_clean_pay.groupby('Payment_Method')
]

# One-Way ANOVA
f_stat, p_val_anova = stats.f_oneway(*groups)

print("=== ONE-WAY ANOVA ===")
print(f"F-Statistic: {f_stat:.4f}")
print(f"p-value:     {p_val_anova:.4e}")

if p_val_anova < 0.05:
  print(
      "Result: Reject Null Hypothesis (H₀) - Transaction amounts differ"
      " significantly by payment method."
  )
else:
  print(
      "Result: Fail to Reject Null Hypothesis (H₀) - No significant difference"
      " in amounts across payment methods."
  )
```

**Phase 6 Statistical Fraud Signals**

```python
import pandas as pd

# Ensure amount z-score and 24h frequency features are calculated
df['Amount_Z'] = (
    df['Transaction_Amount'] - df['Transaction_Amount'].mean()
) / df['Transaction_Amount'].std()
df['Freq_Z'] = (
    df['Number_of_Transactions_Last_24H']
    - df['Number_of_Transactions_Last_24H'].mean()
) / df['Number_of_Transactions_Last_24H'].std()

# Count unique devices used per user across the dataset
device_counts = (
    df.groupby('User_ID')['Device_Used']
    .nunique()
    .reset_index(name='Unique_Devices_Used')
)
df_signals = df.merge(device_counts, on='User_ID', how='left')

# -------------------------------------------------------------------
# Define Statistical Fraud Signals
# -------------------------------------------------------------------

# 1. Sudden High-Value Transfers (Amount Z-Score > 3)
df_signals['Signal_High_Value'] = (df_signals['Amount_Z'] > 3).astype(int)

# 2. Rapid Transaction Bursts (> 15 transactions in 24 hours)
df_signals['Signal_Rapid_Burst'] = (
    df_signals['Number_of_Transactions_Last_24H'] > 15
).astype(int)

# 3. Multiple Device Changes (User has used 3 or more distinct devices)
df_signals['Signal_Multi_Device'] = (
    df_signals['Unique_Devices_Used'] >= 3
).astype(int)

# 4. Late-Night Transaction Spikes (Time between 12:00 AM and 4:00 AM)
df_signals['Signal_Late_Night'] = df_signals['Time_of_Transaction'].isin(
    [0.0, 1.0, 2.0, 3.0, 4.0]
).astype(int)

# 5. New Accounts Making Large Payments (Account Age < 30 Days & Amount > $5,000)
df_signals['Signal_New_Acc_Large_Pay'] = (
    (df_signals['Account_Age'] < 30) & (df_signals['Transaction_Amount'] > 5000)
).astype(int)

# -------------------------------------------------------------------
# Calculate Performance & Fraud Precision per Signal
# -------------------------------------------------------------------
signals_list = [
    ('Sudden High-Value Transfers', 'Signal_High_Value'),
    ('Rapid Transaction Bursts', 'Signal_Rapid_Burst'),
    ('Multiple Device Changes', 'Signal_Multi_Device'),
    ('Late-Night Transaction Spikes', 'Signal_Late_Night'),
    ('New Accounts Making Large Payments', 'Signal_New_Acc_Large_Pay'),
]

report_data = []

for name, col in signals_list:
  total_triggered = df_signals[col].sum()
  fraud_triggered = df_signals[df_signals[col] == 1]['Fraudulent'].sum()
  fraud_rate = (
      (fraud_triggered / total_triggered * 100) if total_triggered > 0 else 0
  )

  report_data.append({
      'Fraud Signal': name,
      'Triggered Count': total_triggered,
      'Fraud Hits': fraud_triggered,
      'Fraud Rate (%)': round(fraud_rate, 2),
  })

# Convert to DataFrame and sort by Fraud Rate
signal_report = pd.DataFrame(report_data).sort_values(
    by='Fraud Rate (%)', ascending=False
)

print("=" * 70)
print("PHASE 6: STATISTICAL FRAUD SIGNAL REPORT")
print("=" * 70)
print(signal_report.to_string(index=False))
```
